# Built-in Data Types & Duck Typing

```python
# Python otomatis mendeteksi tipe data (Duck Typing)
age = 25           # int
ipk = 3.68         # float
name = "DigitalSkola" # str
is_active = True   # bool

print(f"age: {age} -> {type(age)}")
print(f"ipk: {ipk} -> {type(ipk)}")
print(f"name: {name} -> {type(name)}")
print(f"is_active: {is_active} -> {type(is_active)}")
```

# 3. Deep Into Collections

## LIST

```python
# Akses & Konversi List
a = [10, 20, 30]
print("Elemen pertama:", a[0])
print("Elemen terakhir:", a[-1])

# Differences: Append vs Extend
list_a = [1, 2]
list_a.append([3, 4])
print("Result append:", list_a) # Result: [1, 2, [3, 4]]

list_b = [1, 2]
list_b.extend([3, 4])
print("Result extend:", list_b) # Result: [1, 2, 3, 4]
```

## TUPLE

```python
# Inisialisasi & Akses Tuple
tp = ("ID", 101, True)
print("Indeks 1:", tp[1])

# Konversi List ke Tuple
converted_list = tuple(tp)
print("List hasil konversi:", converted_list)

# conversi Tuple ke List
converted_tuple = list(tp)
print("Tuple hasil konversi:", converted_tuple)
```

## Set (Unordered, Unique Values Only)

```python
# Eliminasi Duplikat otomatis
st = {1, 2, 2, 3}
print("Set unik:", st) # Output: {1, 2, 3}

# Update dengan List
st.update([4, 5])
print("Set setelah update list:", st)

# Update dengan number
st.update({4,5,6})
print("Set setelah update number:", st)

# Update dengan String (String diurai menjadi karakter unik)
st_str = {1, 2}
st_str.update('halo')
print("Set setelah update string:", st_str)
```

## Dictionary (Key-Value Pair, Immutable Key, Mutable Value)
Menyimpan data dalam pasangan kata kunci (key) dan nilai (value) untuk *lookup* data secara cepat.

```python
# Membuat & Akses Dictionary
student = {'id': 101, 'name': 'Budi'}
print("Nama siswa:", student['name'])

# Update Data
student['grade'] = 'A'
student.update({'id': 102})
print("Dictionary terupdate:", student)

# Konversi Pasangan Tuple ke Dict
dict_data = dict([
    ('role', 'admin'),
    ('level', 5)
])
print("Dict dari tuple:", dict_data)
```

# Control Flow & Loops
Control Flow mengatur urutan dan kondisi baris kode yang dieksekusi (*conditional statement* dan *looping*).

```python
# Conditional Statement (IF - ELIF - ELSE)

is_loggged_in = True

def show_dashboard():
    print("Welcome to the dashboard!")

if is_loggged_in:
    show_dashboard()
else:
    print("Please log in to access the dashboard.") 


score = 85

if score >= 90:
    print("Grade: A")
elif score >= 75:
    print("Grade: B")
else:
    print("Grade: C")

# For Loop dengan enumerate
items = ['A', 'B', 'C']
print("\n--- Iterasi For Loop ---")
for idx, x in enumerate(items):
    print(f"Index {idx}: {x}")

for x in items:
    print((x))

for x in enumerate(items):
    print((x))

# While Loop
print("\n--- Iterasi While Loop ---")
count = 0
while count < 3:
    print(f"Count: {count}")
    count += 1
```
